```python
import pandas as pd
from supervenn import supervenn
import matplotlib.pyplot as plt

# Load data
icgc = pd.read_excel("../../results/ICGC/ICGC_NetBID2_ms_tab.xlsx")
bcmo = pd.read_excel("../../results/BCMO/BCMO_NetBID2_ms_tab.xlsx")

# Function to get sets by direction only (ignore funcType)
def get_set(df, direction):
    if direction == "pos":
        df_filt = df[
            (df["adj.P.Val.U.Vs.M_DA"] < 0.05) & 
            (df["logFC.U.Vs.M_DA"] > 0) & 
            (df["Size"] > 30)
        ]
    else:  # direction == "neg"
        df_filt = df[
            (df["adj.P.Val.U.Vs.M_DA"] < 0.05) & 
            (df["logFC.U.Vs.M_DA"] < 0) & 
            (df["Size"] > 30)
        ]
    return set(df_filt["originalID"].dropna())

# Define directions and datasets
directions = ["pos", "neg"]
datasets = {"ICGC": icgc, "BCMO": bcmo}

# Build sets dictionary with keys as labels and values as sets of originalID
sets = {}
for ds_name, df in datasets.items():
    for direction in directions:
        key = f"{ds_name}_{direction}"
        sets[key] = get_set(df, direction)

# Prepare lists for supervenn (expects list of sets + labels separately)
sets_list = list(sets.values())
labels = list(sets.keys())

# Plot supervenn
plt.figure(figsize=(10, 6))
supervenn(sets_list, set_annotations=labels)
plt.title("Supervenn: ICGC & BCMO by Positive/Negative")

# Get the current axes and adjust label font size
ax = plt.gca()
for label in ax.get_yticklabels():
    label.set_fontsize(8)
for label in ax.get_xticklabels():
    label.set_fontsize(8)

plt.tight_layout()
plt.savefig("../../figures/SuperVennDiagram.png", dpi=600)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib_venn import venn2, venn2_circles

# Load data
icgc = pd.read_excel("../../results/ICGC/ICGC_NetBID2_ms_tab.xlsx")
bcmo = pd.read_excel("../../results/BCMO/BCMO_NetBID2_ms_tab.xlsx")

# Filter positive drivers
icgc_pos = icgc[(icgc["adj.P.Val.U.Vs.M_DA"] < 0.05) & (icgc["logFC.U.Vs.M_DA"] > 0) & (icgc["Size"] > 30)]
bcmo_pos = bcmo[(bcmo["adj.P.Val.U.Vs.M_DA"] < 0.05) & (bcmo["logFC.U.Vs.M_DA"] > 0) & (bcmo["Size"] > 30)]

# Filter negative drivers
icgc_neg = icgc[(icgc["adj.P.Val.U.Vs.M_DA"] < 0.05) & (icgc["logFC.U.Vs.M_DA"] < 0) & (icgc["Size"] > 30)]
bcmo_neg = bcmo[(bcmo["adj.P.Val.U.Vs.M_DA"] < 0.05) & (bcmo["logFC.U.Vs.M_DA"] < 0) & (bcmo["Size"] > 30)]

# Create sets of originalID
set_icgc_pos = set(icgc_pos['originalID'].dropna())
set_bcmo_pos = set(bcmo_pos['originalID'].dropna())
set_icgc_neg = set(icgc_neg['originalID'].dropna())
set_bcmo_neg = set(bcmo_neg['originalID'].dropna())

# Plot 1x2 figure with Venn diagrams
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Positive Venn
venn2([set_icgc_pos, set_bcmo_pos], set_labels=('ICGC', 'BCMO'), ax=axes[0])
axes[0].set_title('Overlap of Positive Drivers')

# Negative Venn
venn2([set_icgc_neg, set_bcmo_neg], set_labels=('ICGC', 'BCMO'), ax=axes[1])
axes[1].set_title('Overlap of Negative Drivers')

plt.tight_layout()
plt.savefig("../../figures/VennDiagram.png", dpi=600)
plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import auc

# File path
file_path = '../../PASNet/Output/GO/PASNet_pred_ICGC_validation.txt'

# Read the file and parse points
x = []
y = []

with open(file_path, 'r') as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        parts = line.split(',')
        if len(parts) == 2:
            x_val, y_val = map(float, parts)
            x.append(x_val)
            y.append(y_val)

# Sort points by x (False Positive Rate)
sorted_points = sorted(zip(x, y), key=lambda p: p[0])
x_sorted, y_sorted = zip(*sorted_points)

# Calculate AUC
curve_auc = auc(x_sorted, y_sorted)

# Plot ROC curve
plt.plot(x_sorted, y_sorted, marker='o', label=f'AUC = {curve_auc:.4f}')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve from PASNet_pred_ICGC_validation')
plt.legend()
plt.grid(True)
plt.show()
```
