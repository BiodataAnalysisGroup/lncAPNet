```python
def excel_to_gmt(input_xlsx, output_gmt):
    # Load the Excel file
    df = pd.read_excel(input_xlsx, header=None)  # No headers
    
    # Open the output GMT file for writing
    with open(output_gmt, "w") as gmt_file:
        for _, row in df.iterrows():
            # Convert row to tab-separated string
            line = "\t".join(row.dropna().astype(str))  # Remove NaN values
            gmt_file.write(line + "\n")  # Write to GMT file

# Example usage
excel_to_gmt("data/EnrichR/both/GO.xlsx", "data/EnrichR/both/GO.gmt")
print("Conversion complete! GMT file saved.")
```

```python
# Load the Excel files
ms_tab_icgc = pd.read_excel("ICGC_NetBID2/Driver_output/ICGC_NetBID2/DATA/ICGC_NetBID2_ms_tab.xlsx")
ms_tab_bcmo = pd.read_excel("BCMO_NetBID2/Driver_output/BCMO_NetBID2/DATA/BCMO_NetBID2_ms_tab.xlsx")

# Filter the data based on the adjusted p-value column (adj.P.Val.U.Vs.M_DA < 0.05) & Size > 0 & logFC conditions
ms_tab_icgc_pos = ms_tab_icgc[(ms_tab_icgc['adj.P.Val.U.Vs.M_DA'] < 0.05) & (ms_tab_icgc['Size'] > 30)]

ms_tab_bcmo_pos = ms_tab_bcmo[(ms_tab_bcmo['adj.P.Val.U.Vs.M_DA'] < 0.05) & (ms_tab_icgc['Size'] > 30)]

ms_tab_icgc_neg = ms_tab_icgc[(ms_tab_icgc['adj.P.Val.U.Vs.M_DA'] < 0.05) & (ms_tab_icgc['Size'] > 30)]

ms_tab_bcmo_neg = ms_tab_bcmo[(ms_tab_bcmo['adj.P.Val.U.Vs.M_DA'] < 0.05) & (ms_tab_icgc['Size'] > 30)]

# Get unique values from the 'originalID' column
ms_tab_icgc_pos = ms_tab_icgc_pos['originalID'].unique()
ms_tab_bcmo_pos = ms_tab_bcmo_pos['originalID'].unique()
ms_tab_icgc_neg = ms_tab_icgc_neg['originalID'].unique()
ms_tab_bcmo_neg = ms_tab_bcmo_neg['originalID'].unique()

# Find common elements between the two lists and ensure uniqueness
ms_tab_pos = list(set(ms_tab_icgc_pos) & set(ms_tab_bcmo_pos))
ms_tab_neg = list(set(ms_tab_icgc_neg) & set(ms_tab_bcmo_neg))

# Combine the positive and negative results and make sure the result is unique
ms_tab = list(set(ms_tab_pos + ms_tab_neg))  # Concatenate both lists and then ensure uniqueness
ms_tab.sort()  # Sort the final list

# Print the length of the resulting list
print(len(ms_tab))
```

```python
# Load the Excel files
ms_tab_icgc = pd.read_excel("ICGC_NetBID2/Driver_output/ICGC_NetBID2/DATA/ICGC_NetBID2_ms_tab.xlsx")
ms_tab_bcmo = pd.read_excel("BCMO_NetBID2/Driver_output/BCMO_NetBID2/DATA/BCMO_NetBID2_ms_tab.xlsx")

# Filter the data based on the adjusted p-value column (adj.P.Val.U.Vs.M_DA < 0.05) & Size > 0 & logFC conditions
#ms_tab_icgc = ms_tab_icgc[(ms_tab_icgc['adj.P.Val.U.Vs.M_DA'] < 0.05)]
#ms_tab_bcmo = ms_tab_bcmo[(ms_tab_bcmo['adj.P.Val.U.Vs.M_DA'] < 0.05)]

ms_tab_icgc = ms_tab_icgc[(ms_tab_icgc['adj.P.Val.U.Vs.M_DA'] < 0.05) & (ms_tab_icgc['Size'] > 30)]
#ms_tab_bcmo = ms_tab_bcmo[(ms_tab_bcmo['adj.P.Val.U.Vs.M_DA'] < 0.05) & (ms_tab_icgc['Size'] > 30)]

# Get unique values from the 'originalID' column
ms_tab_icgc = ms_tab_icgc['originalID'].unique()
ms_tab_bcmo = ms_tab_bcmo['originalID'].unique()
#ms_tab = ms_tab_icgc
# Find common elements between the two lists and ensure uniqueness
ms_tab = list(set(ms_tab_icgc) & set(ms_tab_bcmo))
ms_tab.sort()

# Print the length of the resulting list
print(len(ms_tab))
```

```python
gene_list = ms_tab

# Run ORA with a custom GMT file
enr = gp.enrichr(
    gene_list=gene_list, 
    gene_sets="data/EnrichR/both/GO.gmt",  # Path to your custom GMT file
    organism="Human", 
    outdir="data/EnrichR/both/GeneOntology_res"
)

results = enr.results

top_300_results = results.sort_values(by="Adjusted P-value", ascending=True).head(300)
top_200_results = results.sort_values(by="Adjusted P-value", ascending=True).head(200)
top_100_results = results.sort_values(by="Adjusted P-value", ascending=True).head(100)

top_300_results = top_300_results[["Term", "Genes"]]

df = top_300_results

top_300_results['Genes'] = top_300_results['Genes'].str.split(';')

top_300_results_n = top_300_results.explode("Genes").pivot_table(index="Term", columns="Genes", aggfunc="size", fill_value=0).reset_index()
top_300_results_n = top_300_results_n.set_index('Term')

top_300_results_n.to_excel("PASNet/Input/GO/pt_fixed.xlsx")
```

```python
# Garbage collection
gc.collect()

working_dir = "/mnt/c/Users/vasileioubill95/Desktop/Projects/lncAPNet_test/"

# Load activity data
activity = pd.read_csv(working_dir + "data/ICGC/activity.csv")
activity['Unnamed: 0'] = activity['Unnamed: 0'].str.replace('_TF', '').str.replace('_SIG', '').str.replace('_LNC', '')
activity = activity.drop_duplicates(subset=['Unnamed: 0']).set_index('Unnamed: 0').T

# Load metadata
metadata = pd.read_csv(working_dir + "data/ICGC/metadata.csv", index_col=0)
metadata['IGHV'] = metadata['IGHV'].replace({'U': '1', 'M': '0'})

# Load pathway data
pt_rest = pd.read_excel(working_dir + "PASNet/Input/REST/pt_fixed.xlsx", index_col=0)
pt_go = pd.read_excel(working_dir + "PASNet/Input/GO/pt_fixed.xlsx", index_col=0)

# Subset activity data
activity_go = activity.loc[:, activity.columns.isin(pt_go.columns)]
activity_rest = activity.loc[:, activity.columns.isin(pt_rest.columns)]

# Merge with metadata
activity_go = activity_go.merge(metadata, left_index=True, right_index=True)
activity_rest = activity_rest.merge(metadata, left_index=True, right_index=True)

# Split into train and test sets
train_go, test_go = train_test_split(activity_go, test_size=0.3, random_state=123, stratify=activity_go['IGHV'])
train_rest, test_rest = train_test_split(activity_rest, test_size=0.3, random_state=123, stratify=activity_rest['IGHV'])

# Save to Excel
train_go.to_excel(working_dir + "PASNet/Input/GO/ICGC_training.xlsx", index=True)
test_go.to_excel(working_dir + "PASNet/Input/GO/ICGC_validation.xlsx", index=True)
train_rest.to_excel(working_dir + "PASNet/Input/REST/ICGC_training.xlsx", index=True)
test_rest.to_excel(working_dir + "PASNet/Input/REST/ICGC_validation.xlsx", index=True)

# BCMO Data Processing
activity = pd.read_csv(working_dir + "data/BCMO/activity.csv")
activity['Unnamed: 0'] = activity['Unnamed: 0'].str.replace('_TF', '').str.replace('_SIG', '').str.replace('_LNC', '')
activity = activity.drop_duplicates(subset=['Unnamed: 0']).set_index('Unnamed: 0').T

# Load BCMO metadata
metadata = pd.read_csv(working_dir + "data/BCMO/metadata.csv", index_col=0)[['IGHV']]
metadata['IGHV'] = metadata['IGHV'].replace({'U': '1', 'M': '0'})

# Subset activity data
activity_go = activity.loc[:, activity.columns.isin(pt_go.columns)]
activity_rest = activity.loc[:, activity.columns.isin(pt_rest.columns)]

# Merge with metadata
activity_go = activity_go.merge(metadata, left_index=True, right_index=True)
activity_rest = activity_rest.merge(metadata, left_index=True, right_index=True)

# Save BCMO test data
activity_go.to_excel(working_dir + "PASNet/Input/GO/BCMO_testing.xlsx", index=True)
activity_rest.to_excel(working_dir + "PASNet/Input/REST/BCMO_testing.xlsx", index=True)
```
