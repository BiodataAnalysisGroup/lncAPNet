```python
import pandas as pd
```

```python
# Load the CSV file
activity_df = pd.read_csv("../../../data/TCGA/activity_ens.csv")
# Load the Excel file
metadata_df = pd.read_excel("../../../data/TCGA/clinical_info.xlsx")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load your feature importance files
shap_validation_go = pd.read_csv("../../PASNet/Output/GO/feature_importance_table.csv", index_col=0)
shap_validation_rest = pd.read_csv("../../PASNet/Output/REST/feature_importance_table.csv", index_col=0)

# Get top 50 genes from each
shap_validation_go_top50 = shap_validation_go.sort_values(by="Importance", ascending=False).head(50)
shap_validation_rest_top50 = shap_validation_rest.sort_values(by="Importance", ascending=False).head(50)

# Combine top 100 genes
top_genes = pd.concat([shap_validation_go_top50, shap_validation_rest_top50]).index.tolist()

# Subset your activity dataframe to these top genes
activity_top100 = activity_df[top_genes]

# Subset metadata for the annotation
metadata_subset = metadata_df[['Status', 'Status_stages', 'ajcc_pathologic_n']]

# Make sure the activity dataframe columns match metadata rows
# Assuming activity_df columns are sample IDs
metadata_subset = metadata_subset.loc[activity_top100.columns]

# Create the heatmap
plt.figure(figsize=(20, 10))
sns.heatmap(
    activity_top100,
    cmap="vlag",            # Diverging colormap
    xticklabels=True,
    yticklabels=True,
    cbar_kws={'label': 'Expression/Activity'}
)

# Optionally, add column annotations
# seaborn clustermap supports annotations better
sns.clustermap(
    activity_top100,
    col_colors=pd.get_dummies(metadata_subset),  # Convert categorical to colors
    cmap="vlag",
    figsize=(20, 15)
)

plt.show()
```
