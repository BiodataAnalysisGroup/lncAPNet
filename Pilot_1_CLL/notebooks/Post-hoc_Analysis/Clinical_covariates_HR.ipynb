```python
def kaplan_meier_plots(
    clinical_file,
    expr_file,
    genes,
    output_file,
    split_method="median",
    fig_title=None,
    time_col="OS.from.sampling"
):
    """
    Generate Kaplan-Meier survival plots for a list of genes.

    Parameters:
    - clinical_file: path to clinical CSV (must have Sample_ID and Status)
    - expr_file: path to expression/activity CSV (genes x samples)
    - genes: list of gene names to analyze
    - output_file: path to save the figure
    - split_method: "median" or "quartile"
    - fig_title: string for the figure-wide title (optional)
    - time_col: name of the time column in clinical data
    """
    # 1. Load clinical data
    clinical = pd.read_csv(clinical_file)
    clinical = clinical.rename(columns={"died": "Status"})
    
    # 2. Load expression/activity matrix
    expr = pd.read_csv(expr_file, index_col=0)
    
    # 3. Determine subplot grid size dynamically
    n_genes = len(genes)
    n_cols = 2
    n_rows = math.ceil(n_genes / n_cols)
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(16, 9 * n_rows))
    axes = axes.flatten()
    
    # 4. Figure-wide title
    if fig_title is None:
        title_map = {
            "median": "Survival Analysis of BCMO for ALL Genes -- ACTIVITY",
            "quartile": "Survival Analysis of BCMO for Quartile Genes (Activity)"
        }
        fig_title = title_map.get(split_method, "Survival Analysis")
    fig.suptitle(fig_title, fontsize=30, y=1.02)
    
    # 5. Loop through genes
    for i, gene in enumerate(genes):
        ax = axes[i]
        
        if gene not in expr.index:
            ax.text(0.5, 0.5, f"{gene} not found", ha='center', va='center', fontsize=14)
            ax.axis('off')
            continue
        
        # Extract gene activity
        gene_activity = expr.loc[gene].rename("Activity")
        
        # Merge with clinical data
        df = clinical.merge(gene_activity, left_on="Sample_ID", right_index=True)
        
        # Define High/Low groups
        if split_method == "median":
            df[f"{gene}_group"] = (df["Activity"] > df["Activity"].median()).map({True: "High", False: "Low"})
            df_to_use = df
        elif split_method == "quartile":
            q_low = df["Activity"].quantile(0.25)
            q_high = df["Activity"].quantile(0.75)
            df_to_use = df[(df["Activity"] <= q_low) | (df["Activity"] >= q_high)].copy()
            df_to_use[f"{gene}_group"] = df_to_use["Activity"].apply(lambda x: "High" if x >= q_high else "Low")
        else:
            raise ValueError("split_method must be 'median' or 'quartile'")
        
        # Kaplan–Meier plot
        kmf = KaplanMeierFitter()
        for group, grouped_df in df_to_use.groupby(f"{gene}_group"):
            kmf.fit(grouped_df[time_col], grouped_df["Status"], label=group)
            kmf.plot_survival_function(ax=ax, ci_show=True)
        
        # Log-rank test
        high = df_to_use[df_to_use[f"{gene}_group"] == "High"]
        low = df_to_use[df_to_use[f"{gene}_group"] == "Low"]
        results = logrank_test(
            high[time_col], low[time_col],
            event_observed_A=high["Status"], event_observed_B=low["Status"]
        )
        pval = results.p_value

        # Compute hazard ratio with CoxPH
        df_cph = df_to_use[[time_col, "Status", f"{gene}_group"]].copy()
        df_cph[f"{gene}_group"] = df_cph[f"{gene}_group"].map({"Low": 0, "High": 1})  # numeric encoding
        
        cph = CoxPHFitter()
        try:
            cph.fit(df_cph, duration_col=time_col, event_col="Status")
            hr = cph.hazard_ratios_[f"{gene}_group"]
        except Exception:
            hr = float("nan")
        
        # Format values
        if pval >= 0.005:
            pval_str = f"{pval:.4f}"
        else:
            pval_str = f"{pval:.1e}".replace("e-0", "e-").replace("e", "×10")
        
        hr_str = f"{hr:.2f}" if not np.isnan(hr) else "NA"
        
        # Set subplot title and labels
        ax.set_title(f"{gene} (HR={hr_str}, p={pval_str})", fontsize=25)
        ax.set_xlabel(f"Time from Sampling ({time_col})", fontsize=16)
        ax.set_ylabel("Overall Survival Probability", fontsize=16)
        ax.tick_params(axis='both', labelsize=14)
        ax.legend(fontsize=14)
    
    # Remove unused axes if any
    for j in range(i + 1, len(axes)):
        axes[j].axis('off')
    
    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.savefig(output_file, dpi=300, bbox_inches='tight')
    plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from lifelines import KaplanMeierFitter, CoxPHFitter
from lifelines.statistics import logrank_test

# --------------------------
# Step 1: Load data
# --------------------------
clinical = pd.read_csv("../../data/BCMO/clinical_info.csv")
expr = pd.read_csv("../../data/BCMO/count_matrix.csv", index_col=0)  # genes as rows

# RNAs of interest
rna_of_interest = ["FIRRE", "LINC00158", "KCNJ2-AS1", "MIR4458HG", "FLNB-AS1"]

# Subset expression matrix
expr = expr.loc[rna_of_interest].T
expr.index.name = "Sample_ID"

# Merge with clinical data
data = clinical.merge(expr, on="Sample_ID")

# --------------------------
# Step 2: Create combined score and High/Low group
# --------------------------
data["combined_score"] = data[rna_of_interest].sum(axis=1)
data["combined_group"] = np.where(data["combined_score"] > data["combined_score"].median(), "High", "Low")

# --------------------------
# Step 3: Kaplan-Meier plot for combined score
# --------------------------
# Log-rank test
high = data[data["combined_group"]=="High"]
low = data[data["combined_group"]=="Low"]
results = logrank_test(high["OS.from.sampling"], low["OS.from.sampling"],
                       event_observed_A=high["died"], event_observed_B=low["died"])
pval = results.p_value

# Cox PH for hazard ratio
cox_df = data[["OS.from.sampling", "died", "combined_group"]].copy()
cox_df["combined_group_num"] = cox_df["combined_group"].map({"Low":0, "High":1})

cph = CoxPHFitter()
cph.fit(cox_df[["OS.from.sampling", "died", "combined_group_num"]],
        duration_col="OS.from.sampling", event_col="died")

hr = cph.hazard_ratios_["combined_group_num"]
hr_ci = cph.confidence_intervals_.loc["combined_group_num"].tolist()
hr_ci_str = f"{hr_ci[0]:.2f}-{hr_ci[1]:.2f}"

# Plot KM curve
kmf = KaplanMeierFitter()
fig, ax = plt.subplots(figsize=(8,6))
for label, group in data.groupby("combined_group"):
    kmf.fit(group["OS.from.sampling"], group["died"], label=label)
    kmf.plot_survival_function(ax=ax)

plt.title(f"Expression -- KM survival by combined 5 lncRNA score\nHR={hr:.2f} (95% CI: {hr_ci_str}), p={pval:.3f}", fontsize=14)
plt.xlabel("Time from Sampling")
plt.ylabel("Survival Probability")
plt.tight_layout()
plt.savefig("../../figures/KaplanMeier_OS_combined_lncRNAs_BCMO_expression.png", dpi=300, bbox_inches='tight')
plt.show()

# --------------------------
# Step 4: Optional multivariate Cox regression on individual RNAs
# --------------------------
cox_df_multi = data[["OS.from.sampling", "died"] + rna_of_interest].copy()
cph_multi = CoxPHFitter()
cph_multi.fit(cox_df_multi, duration_col="OS.from.sampling", event_col="died")
cph_multi.print_summary()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math
from lifelines import KaplanMeierFitter, CoxPHFitter
from lifelines.statistics import logrank_test

def kaplan_meier_plots(
    clinical_file,
    expr_file,
    genes,
    output_file,
    split_method="median",
    fig_title=None,
    time_col="OS.from.sampling"
):
    """
    Generate Kaplan-Meier survival plots with hazard ratios for a list of genes.

    Parameters:
    - clinical_file: path to clinical CSV (must have Sample_ID and Status)
    - expr_file: path to expression/activity CSV (genes x samples)
    - genes: list of gene names to analyze
    - output_file: path to save the figure
    - split_method: "median" or "quartile"
    - fig_title: string for the figure-wide title (optional)
    - time_col: name of the time column in clinical data
    """
    # 1. Load clinical data
    clinical = pd.read_csv(clinical_file)
    clinical = clinical.dropna(subset=['donor_vital_status', 'donor_survival_time'])
    clinical['donor_vital_status'] = clinical['donor_vital_status'].map({'alive': 0, 'deceased': 1})
    clinical = clinical.rename(columns={"donor_vital_status": "Status", "Sample_ID": "Samples"})
    
    # 2. Load expression/activity matrix
    expr = pd.read_csv(expr_file, index_col=0)
    
    # 3. Determine subplot grid size dynamically
    n_genes = len(genes)
    n_cols = 2
    n_rows = math.ceil(n_genes / n_cols)
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(16, 9 * n_rows))
    axes = axes.flatten()
    
    # 4. Figure-wide title
    if fig_title is None:
        title_map = {
            "median": "Survival Analysis of ICGC for ALL Genes -- ACTIVITY",
            "quartile": "Survival Analysis of ICGC for Quartile Genes (Activity)"
        }
        fig_title = title_map.get(split_method, "Survival Analysis")
    fig.suptitle(fig_title, fontsize=30, y=1.02)
    
    # 5. Loop through genes
    for i, gene in enumerate(genes):
        ax = axes[i]
        
        if gene not in expr.index:
            ax.text(0.5, 0.5, f"{gene} not found", ha='center', va='center', fontsize=14)
            ax.axis('off')
            continue
        
        # Extract gene activity
        gene_activity = expr.loc[gene].rename("Activity")
        
        # Merge with clinical data
        df = clinical.merge(gene_activity, left_on="Samples", right_index=True)
        
        # Define High/Low groups
        if split_method == "median":
            df[f"{gene}_group"] = (df["Activity"] > df["Activity"].median()).map({True: "High", False: "Low"})
            df_to_use = df
        elif split_method == "quartile":
            q_low = df["Activity"].quantile(0.25)
            q_high = df["Activity"].quantile(0.75)
            df_to_use = df[(df["Activity"] <= q_low) | (df["Activity"] >= q_high)].copy()
            df_to_use[f"{gene}_group"] = df_to_use["Activity"].apply(lambda x: "High" if x >= q_high else "Low")
        else:
            raise ValueError("split_method must be 'median' or 'quartile'")
        
        # Kaplan–Meier plot
        kmf = KaplanMeierFitter()
        for group, grouped_df in df_to_use.groupby(f"{gene}_group"):
            kmf.fit(grouped_df[time_col], grouped_df["Status"], label=group)
            kmf.plot_survival_function(ax=ax, ci_show=True)
        
        # Log-rank test
        high = df_to_use[df_to_use[f"{gene}_group"] == "High"]
        low = df_to_use[df_to_use[f"{gene}_group"] == "Low"]
        results = logrank_test(
            high[time_col], low[time_col],
            event_observed_A=high["Status"], event_observed_B=low["Status"]
        )
        pval = results.p_value

        # Cox PH for hazard ratio
        df_to_use_cox = df_to_use[[time_col, "Status", f"{gene}_group"]].copy()
        df_to_use_cox[f"{gene}_group"] = df_to_use_cox[f"{gene}_group"].map({"Low": 0, "High": 1})
        cph = CoxPHFitter()
        cph.fit(df_to_use_cox, duration_col=time_col, event_col="Status")
        hr = cph.hazard_ratios_[f"{gene}_group"]

        # Format p-value and HR
        if pval >= 0.005:
            pval_str = f"{pval:.4f}"
        else:
            pval_str = f"{pval:.1e}".replace("e-0", "e-").replace("e", "×10")
        hr_str = f"{hr:.2f}"

        # Set subplot title and labels
        ax.set_title(f"{gene} (HR={hr_str}, p={pval_str})", fontsize=25)
        ax.set_xlabel(f"Time from Sampling ({time_col})", fontsize=16)
        ax.set_ylabel("Overall Survival Probability", fontsize=16)
        ax.tick_params(axis='both', labelsize=14)
        ax.legend(fontsize=14)
    
    # Remove unused axes if any
    for j in range(i + 1, len(axes)):
        axes[j].axis('off')
    
    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.savefig(output_file, dpi=300, bbox_inches='tight')
    plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from lifelines import KaplanMeierFitter, CoxPHFitter
from lifelines.statistics import logrank_test

# --------------------------
# Step 1: Load data
# --------------------------
clinical = pd.read_csv("../../data/ICGC/clinical_info.csv")
expr = pd.read_csv("../../data/ICGC/activity_matrix_pvalfilt.csv", index_col=0)  # genes as rows

# RNAs of interest
rna_of_interest = ["FIRRE_NC", "LINC00158_NC", "KCNJ2-AS1_NC", "MIR4458HG_NC", "FLNB-AS1_NC"]

# Subset expression matrix and transpose
expr = expr.loc[rna_of_interest].T
expr.index.name = "Samples"

# Merge with clinical data
data = clinical.merge(expr, on="Samples")

# --------------------------
# Step 2: Preprocess clinical data
# --------------------------
# Convert donor_vital_status to numeric (alive=0, deceased=1)
data["donor_vital_status_num"] = data["donor_vital_status"].map({"alive": 0, "deceased": 1})

# Ensure survival time is numeric and drop rows with missing values
data["donor_survival_time"] = pd.to_numeric(data["donor_survival_time"], errors='coerce')
data = data.dropna(subset=["donor_survival_time", "donor_vital_status_num"] + rna_of_interest)

# --------------------------
# Step 3: Create combined score and High/Low group
# --------------------------
data["combined_score"] = data[rna_of_interest].sum(axis=1)
data["combined_group"] = np.where(
    data["combined_score"] > data["combined_score"].median(), "High", "Low"
)

# --------------------------
# Step 4: Kaplan-Meier plot for combined score
# --------------------------
# Log-rank test
high = data[data["combined_group"] == "High"]
low = data[data["combined_group"] == "Low"]

results = logrank_test(
    high["donor_survival_time"], low["donor_survival_time"],
    event_observed_A=high["donor_vital_status_num"],
    event_observed_B=low["donor_vital_status_num"]
)
pval = results.p_value

# Cox PH for hazard ratio
cox_df = data[["donor_survival_time", "donor_vital_status_num", "combined_group"]].copy()
cox_df["combined_group_num"] = cox_df["combined_group"].map({"Low": 0, "High": 1})

cph = CoxPHFitter()
cph.fit(
    cox_df[["donor_survival_time", "donor_vital_status_num", "combined_group_num"]],
    duration_col="donor_survival_time",
    event_col="donor_vital_status_num"
)

hr = cph.hazard_ratios_["combined_group_num"]
hr_ci = cph.confidence_intervals_.loc["combined_group_num"].tolist()
hr_ci_str = f"{hr_ci[0]:.2f}-{hr_ci[1]:.2f}"

# Format p-value: scientific if < 0.001, otherwise 3 decimals
if pval < 0.001:
    pval_str = f"{pval:.1e}"   # e.g. 5.4e-05
else:
    pval_str = f"{pval:.3f}"

# Plot KM curve
kmf = KaplanMeierFitter()
fig, ax = plt.subplots(figsize=(8, 6))
for label, group in data.groupby("combined_group"):
    kmf.fit(group["donor_survival_time"], group["donor_vital_status_num"], label=label)
    kmf.plot_survival_function(ax=ax)

plt.title(
    f"Activity -- KM survival by combined 5 lncRNA score\n"
    f"HR={hr:.2f} (95% CI: {hr_ci_str}), p={pval_str}",
    fontsize=14
)
plt.xlabel("Time from Sampling")
plt.ylabel("Survival Probability")
plt.tight_layout()
plt.savefig("../../figures/KaplanMeier_OS_combined_lncRNAs_ICGC_activity.png", dpi=300, bbox_inches='tight')
plt.show()

# --------------------------
# Step 5: Optional multivariate Cox regression on individual RNAs
# --------------------------
cox_df_multi = data[["donor_survival_time", "donor_vital_status_num"] + rna_of_interest].copy()
cph_multi = CoxPHFitter()
cph_multi.fit(
    cox_df_multi,
    duration_col="donor_survival_time",
    event_col="donor_vital_status_num"
)
cph_multi.print_summary()
```

```python
def kaplan_meier_plots(
    clinical_file,
    expr_file,
    genes,
    output_file,
    split_method="median",
    fig_title=None,
    time_col="OS.from.sampling",
    status=None
):
    """
    Generate Kaplan-Meier survival plots for a list of genes.

    Parameters:
    - clinical_file: path to clinical CSV (must have Sample_ID and Status)
    - expr_file: path to expression/activity CSV (genes x samples)
    - genes: list of gene names to analyze
    - output_file: path to save the figure
    - split_method: "median" or "quartile"
    - fig_title: string for the figure-wide title (optional)
    - time_col: name of the time column in clinical data
    """
    # 1. Load clinical data
    clinical = pd.read_csv(clinical_file)
    clinical = clinical.rename(columns={"died": "Status"})
    clinical = clinical[clinical['IGHV'] == status]
    
    # 2. Load expression/activity matrix
    expr = pd.read_csv(expr_file, index_col=0)
    
    # 3. Determine subplot grid size dynamically
    n_genes = len(genes)
    n_cols = 2
    n_rows = math.ceil(n_genes / n_cols)
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(16, 9 * n_rows))
    axes = axes.flatten()
    
    # 4. Figure-wide title
    if fig_title is None:
        title_map = {
            "median": "Survival Analysis of BCMO for ALL Genes -- ACTIVITY",
            "quartile": "Survival Analysis of BCMO for Quartile Genes (Activity)"
        }
        fig_title = title_map.get(split_method, "Survival Analysis")
    fig.suptitle(fig_title, fontsize=30, y=1.02)
    
    # 5. Loop through genes
    for i, gene in enumerate(genes):
        ax = axes[i]
        
        if gene not in expr.index:
            ax.text(0.5, 0.5, f"{gene} not found", ha='center', va='center', fontsize=14)
            ax.axis('off')
            continue
        
        # Extract gene activity
        gene_activity = expr.loc[gene].rename("Activity")
        
        # Merge with clinical data
        df = clinical.merge(gene_activity, left_on="Sample_ID", right_index=True)
        
        # Define High/Low groups
        if split_method == "median":
            df[f"{gene}_group"] = (df["Activity"] > df["Activity"].median()).map({True: "High", False: "Low"})
            df_to_use = df
        elif split_method == "quartile":
            q_low = df["Activity"].quantile(0.25)
            q_high = df["Activity"].quantile(0.75)
            df_to_use = df[(df["Activity"] <= q_low) | (df["Activity"] >= q_high)].copy()
            df_to_use[f"{gene}_group"] = df_to_use["Activity"].apply(lambda x: "High" if x >= q_high else "Low")
        else:
            raise ValueError("split_method must be 'median' or 'quartile'")
        
        # Kaplan–Meier plot
        kmf = KaplanMeierFitter()
        for group, grouped_df in df_to_use.groupby(f"{gene}_group"):
            kmf.fit(grouped_df[time_col], grouped_df["Status"], label=group)
            kmf.plot_survival_function(ax=ax, ci_show=True)
        
        # Log-rank test
        high = df_to_use[df_to_use[f"{gene}_group"] == "High"]
        low = df_to_use[df_to_use[f"{gene}_group"] == "Low"]
        results = logrank_test(
            high[time_col], low[time_col],
            event_observed_A=high["Status"], event_observed_B=low["Status"]
        )
        pval = results.p_value

        # Compute hazard ratio with CoxPH
        df_cph = df_to_use[[time_col, "Status", f"{gene}_group"]].copy()
        df_cph[f"{gene}_group"] = df_cph[f"{gene}_group"].map({"Low": 0, "High": 1})  # numeric encoding
        
        cph = CoxPHFitter()
        try:
            cph.fit(df_cph, duration_col=time_col, event_col="Status")
            hr = cph.hazard_ratios_[f"{gene}_group"]
        except Exception:
            hr = float("nan")
        
        # Format values
        if pval >= 0.005:
            pval_str = f"{pval:.4f}"
        else:
            pval_str = f"{pval:.1e}".replace("e-0", "e-").replace("e", "×10")
        
        hr_str = f"{hr:.2f}" if not np.isnan(hr) else "NA"
        
        # Set subplot title and labels
        ax.set_title(f"{gene} (HR={hr_str}, p={pval_str})", fontsize=25)
        ax.set_xlabel(f"Time from Sampling ({time_col})", fontsize=16)
        ax.set_ylabel("Overall Survival Probability", fontsize=16)
        ax.tick_params(axis='both', labelsize=14)
        ax.legend(fontsize=14)
    
    # Remove unused axes if any
    for j in range(i + 1, len(axes)):
        axes[j].axis('off')
    
    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.savefig(output_file, dpi=300, bbox_inches='tight')
    plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from lifelines import KaplanMeierFitter, CoxPHFitter
from lifelines.statistics import logrank_test

# --------------------------
# Helper function to create KM curve
# --------------------------
def km_curve(data, rna_cols, time_col, event_col, title, ax):
    # Combined score & group
    data["combined_score"] = data[rna_cols].sum(axis=1)
    data["combined_group"] = np.where(data["combined_score"] > data["combined_score"].median(), "High", "Low")

    # Log-rank test
    high = data[data["combined_group"]=="High"]
    low = data[data["combined_group"]=="Low"]
    results = logrank_test(high[time_col], low[time_col],
                           event_observed_A=high[event_col],
                           event_observed_B=low[event_col])
    pval = results.p_value
    pval_str = f"{pval:.3f}" if pval >= 0.001 else f"{pval:.1e}"

    # Cox PH for HR
    cox_df = data[[time_col, event_col, "combined_group"]].copy()
    cox_df["combined_group_num"] = cox_df["combined_group"].map({"Low":0, "High":1})
    cph = CoxPHFitter()
    cph.fit(cox_df[[time_col, event_col, "combined_group_num"]],
            duration_col=time_col, event_col=event_col)
    hr = cph.hazard_ratios_["combined_group_num"]
    hr_ci = cph.confidence_intervals_.loc["combined_group_num"].tolist()
    hr_ci_str = f"{hr_ci[0]:.2f}-{hr_ci[1]:.2f}"

    # Plot KM
    kmf = KaplanMeierFitter()
    for label, group in data.groupby("combined_group"):
        kmf.fit(group[time_col], group[event_col], label=label)
        kmf.plot_survival_function(ax=ax)
    
    ax.set_title(f"{title}\nHR={hr:.2f} (95% CI: {hr_ci_str}), p={pval_str}", fontsize=12)
    ax.set_xlabel("Time from Sampling")
    ax.set_ylabel("Survival Probability")

# --------------------------
# Step 1: Load datasets
# --------------------------
# BCMO
clinical_bcmo = pd.read_csv("../../../data/BCMO/clinical_info.csv")
expr_bcmo = pd.read_csv("../../../data/BCMO/activity_matrix_pvalfilt.csv", index_col=0)
rna_bcmo = ["FIRRE_NC", "LINC00158_NC", "KCNJ2-AS1_NC", "MIR4458HG_NC", "FLNB-AS1_NC"]
expr_bcmo = expr_bcmo.loc[rna_bcmo].T
expr_bcmo.index.name = "Sample_ID"
data_bcmo = clinical_bcmo.merge(expr_bcmo, on="Sample_ID")

# ICGC
clinical_icgc = pd.read_csv("../../../data/ICGC/clinical_info.csv")
expr_icgc = pd.read_csv("../../../data/ICGC/activity_matrix_pvalfilt.csv", index_col=0)
rna_icgc = ["FIRRE_NC", "LINC00158_NC", "KCNJ2-AS1_NC", "MIR4458HG_NC", "FLNB-AS1_NC"]
expr_icgc = expr_icgc.loc[rna_icgc].T
expr_icgc.index.name = "Samples"
data_icgc = clinical_icgc.merge(expr_icgc, on="Samples")

# Preprocess ICGC clinical
data_icgc["donor_vital_status_num"] = data_icgc["donor_vital_status"].map({"alive":0, "deceased":1})
data_icgc["donor_survival_time"] = pd.to_numeric(data_icgc["donor_survival_time"], errors='coerce')
data_icgc = data_icgc.dropna(subset=["donor_survival_time", "donor_vital_status_num"] + rna_icgc)

# --------------------------
# Step 2: Plot 1x2 multipanel
# --------------------------
fig, axes = plt.subplots(1, 2, figsize=(14,6), sharey=True)

km_curve(data_bcmo, rna_bcmo, "OS.from.sampling", "died", "BCMO Activity", axes[0])
km_curve(data_icgc, rna_icgc, "donor_survival_time", "donor_vital_status_num", "ICGC Activity", axes[1])

plt.tight_layout()
plt.savefig("../../figures/KM_BCMO_ICGC_activity_1x2.png", dpi=300, bbox_inches='tight')
plt.show()
```
